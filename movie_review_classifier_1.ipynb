```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis")
```

```python
movie_reviews = [
    "This film was an absolute masterpiece from start to finish.",
    "I fell asleep halfway through, it was so boring.",
    "The acting was wooden and the plot made no sense.",
    "A stunning visual experience with a powerful soundtrack.",
    "Waste of two hours I will never get back.",
    "The chemistry between the leads made this movie unforgettable.",
    "Predictable ending, but the journey there was fun.",
    "One of the worst sequels I have ever seen.",
    "Brilliant direction and a genuinely moving story.",
    "Overhyped and disappointing, I expected much more."
]

results = classifier(movie_reviews)

for review, result in zip(movie_reviews, results):
    print(f"Review: {review}")
    print(f"  -> {result['label']} (score: {result['score']:.4f})\n")
```

```python
log = [
    {"review": r, "label": res["label"], "score": round(res["score"], 4)}
    for r, res in zip(movie_reviews, results)
]

for entry in log:
    print(entry)
```
